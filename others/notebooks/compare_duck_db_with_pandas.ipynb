### DuckDB vs Pandas for READ operations

```python
import duckdb
import pandas as pd
```

```python
# Pandas processes the entire database inside the computer's RAM memory. 
# For large databases, it can take up a big amount of the local's machine space. 

df = pd.read_csv('/Users/julianasampar/Desktop/learning_dev/personal_dev/pyprojects/others/archive/Amazon Customer Behavior Survey.csv')
display(df)
```

```python
# DuckDB is a database that allows to read massive databases and faster.
# It identifies the chunk of local memory space and automatically stream the process into batches.
# DuckDB can read csv, parquet and etc files and perform SQL queries on top of it.

db = duckdb.sql("SELECT * FROM '/Users/julianasampar/Desktop/learning_dev/personal_dev/pyprojects/others/archive/Amazon Customer Behavior Survey.csv' ")
display(db)
```

```python
# We can transform the db result into Pandas Dataframe

df_db = db.df()
display(df_db)
```

```python
# DuckDB can also read Pandas dataframes
db_df = duckdb.sql("SELECT * FROM df_db")
display(db_df)
```

### DuckDB vs SQLite for Database Creation

```python
import sqlite3
```

```python
connection = sqlite3.connect('/Users/julianasampar/Desktop/learning_dev/personal_dev/pyprojects/dbt_database.db')
cursor = connection.cursor()

cursor.execute("SELECT * FROM dim_films LIMIT 1")
print(cursor.fetchall())

connection.commit()
cursor.close()
connection.close()
```

```python
connection = duckdb.connect('dbt_database.db')
connection.sql("CREATE OR REPLACE TABLE int_rentals (id INT)")
print(connection.sql("SELECT * FROM int_rentals"))
connection.commit()
connection.close()
```
