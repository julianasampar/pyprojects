```python
from desktop_notifier import DesktopNotifier, DEFAULT_SOUND
import asyncio

title = "TIME FOR THE NEWS!!!!"

content = "You were able to make this function work!!!"

def schedule_notification(title:str, content:str):
    
    notifier = DesktopNotifier()
    
    async def main(notifier):
        await notifier.send(
            title=title,
            message=content,
            timeout=0.5,
            sound=DEFAULT_SOUND,
        )

    return  asyncio.run(main(notifier))

schedule_notification(title=title, content=content)
```
